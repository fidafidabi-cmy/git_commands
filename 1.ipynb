```python
con=sqlite3.connect('demo.db')
cursor=con.cursor()
cursor.execute("""
create table if not exists orders(
orderid integer primary key,order_date date,product text,price real)
""")
con.commit()
cursor.close()
con.close()
```

```python
con=sqlite3.connect('demo.db')
cursor=con.cursor()
cursor.execute("""
insert into orders(orderid,order_date,product)
values(1,'2026-11-24','laptop')
""")
con.commit()
cursor.close()
con.close()
```
